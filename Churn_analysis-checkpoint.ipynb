```python
conn = sqlite3.connect('customer_churn.db')

sql_query = """
SELECT name
FROM sqlite_master
WHERE type='table'
"""

tables = pd.read_sql(sql_query, conn)

for table_name in tables['name']:
    df = pd.read_sql(f"SELECT * FROM {table_name}", conn)
    globals()[f"df_{table_name}"] = df
    print(f"Created dataframe: df_{table_name}")

conn.close()
```

```python
churn_rate = df['churn_flag'].mean()*100
print("Churn Rate = ", round(churn_rate,2), "%")
```

```python
retention_rate = 100- churn_rate
print("Retention Rate = ", round(retention_rate,2), "%")
```

```python
# 3. Churn by plan type

churn_by_plan = (df.groupby('plan_type')['churn_flag'].mean().mul(100).round(2).reset_index(name= 'churn_rate_pct'))
print(churn_by_plan)
```

```python
# 4a. Churn by state + sum(revenue) & count of users

churn_by_state = df.groupby('state').agg(
    churn_rate=('churn_flag', 'mean'),
    total_revenue=('monthly_charges', 'sum'),
    user_count=('customerid', 'count')
).reset_index()

churn_by_state['churn_rate'] = (
    churn_by_state['churn_rate'] * 100
).round(2)

churn_by_state = churn_by_state.sort_values(
    'churn_rate',
    ascending=False
)

churn_by_state['total_revenue'] = (
    churn_by_state['total_revenue'].round(2)
)

churn_by_state
```

```python
# 4 b. Churn by subscription type + sum(revenue) & count of users

churn_by_subscription = df.groupby('subscription_type').agg(
    churn_rate=('churn_flag', 'mean'),
    total_revenue=('monthly_charges', 'sum'),
    user_count=('customerid', 'count')
).reset_index()

churn_by_subscription['churn_rate'] = (
    churn_by_subscription['churn_rate'] * 100
).round(2)

churn_by_subscription

churn_by_subscription = churn_by_subscription.sort_values(
    'churn_rate',
    ascending=False
)

churn_by_subscription

```
